```python
#membersihkan data normalized-loses
#mencari jumlah records yang memiliki nilai '?' pada normalized losses
df['normalized-losses'].loc[df['normalized-losses'] == '?'].count()
```

```python
#mengatur nilai missing value menjadi rata-rata normalized-loses dan mengubah tipe data menjadi integer
nl = df['normalized-losses'].loc[df['normalized-losses'] != '?']
nlmean = nl.astype(str).astype(int).mean()
df['normalized-losses'] = df['normalized-losses'].replace('?',nlmean).astype(int)
df['normalized-losses'].head()
```

```python
#membersihkan data kolom price
#mencari jumlah data yang bukan numeric
df['price'].str.isnumeric().value_counts()
```

```python
#list nilai-nilai yang bukan numeric
df['price'].loc[df['price'].str.isnumeric() == False]
```

```python
#mengatur nilai missing value menjadi rata-rata price dan mengubah tipe data menjadi integer
price = df['price'].loc[df['price'] != '?']
pmean = price.astype(str).astype(int).mean()
df['price'] = df['price'].replace('?',pmean).astype(int)
df['price'].head()
```

```python
#membersihkan data kolom horsepower
#memeriksa angka dan mengganti dengan nilai rata-rata dan mengubah tipe data menjadi integer
df['horsepower'].str.isnumeric().value_counts()
horsepower = df['horsepower'].loc[df['horsepower'] != '?']
hpmean = horsepower.astype(str).astype(int).mean()
df['horsepower'] = df['horsepower'].replace('?',pmean).astype(int)
```

```python
#memeriksa outlier pada horsepower
df.loc[df['horsepower'] > 10000]
```

```python
#mengeluarkan data outlier pada horsepower
df[np.abs(df.horsepower-df.horsepower.mean())<=(3*df.horsepower.std())]
```

```python
#membersihkan data kolom bore
# mencari invalid value
df['bore'].loc[df['bore'] == '?']
```

```python
# ganti yang bukan numeric ke null dan konversi tipe datanya
df['bore'] = pd.to_numeric(df['bore'],errors='coerce') #Used to determine how to handle values ​​that cannot be converted to numeric
df.dtypes
```

```python
#membersihkan data stroke
#mengganti yang bukan angka ke null dan to null dan konversi tipe datanya
df['stroke'] = pd.to_numeric(df['stroke'],errors='coerce')
df.dtypes
```

```python
#membersihkan data peak rpm
#konversi yang bukan numeric ke null dan konversi tipe datanya
df['peak-rpm'] = pd.to_numeric(df['peak-rpm'],errors='coerce')
df.dtypes
```

```python
#membersihkan data num-of-doors
#hapus yang mempunyai nilai '2'
df['num-of-doors'].loc[df['num-of-doors'] == '?']
df= df[df['num-of-doors'] != '?']
df['num-of-doors'].loc[df['num-of-doors'] == '?']
```

```python
pivot_table = df.pivot_table(index='fuel-type', columns='aspiration', aggfunc='size', fill_value=0)
pivot_table
```

```python
ax = pivot_table.plot(kind='bar', figsize=(10, 6))
plt.title("Diagram Batang Jenis Bahan Bakar dan Aspirasi")
plt.xlabel('Jenis Bahan Bakar')
plt.ylabel('Jumlah Bahan Bakar')
plt.xticks(rotation=0)

# Menampilkan label pada tiap grafik batang
for i in range(len(ax.patches)):
    ax.text(ax.patches[i].get_x() + ax.patches[i].get_width() / 2, ax.patches[i].get_height(),
            str(ax.patches[i].get_height()), ha='center', va='bottom')

# Menampilkan diagram batang
plt.show()
```

```python
# Menghitung jumlah jenis BBM terbanyak
jenis_bbm_terbanyak = df[index='fuel-type', columns='aspiration'].value_counts().idxmax()

# Menghitung jumlah masing-masing jenis BBM
jumlah_bbm = df['fuel-type'].value_counts()

# Membuat diagram batang
plt.figure(figsize=(10, 6))  # Atur ukuran diagram sesuai kebutuhan
plt.bar(jumlah_bbm.index, jumlah_bbm.values)
plt.xlabel('Jenis BBM')
plt.ylabel('Jumlah')
plt.title(f'Jenis BBM Terbanyak: {jenis_bbm_terbanyak}')
plt.xticks(rotation=45)
plt.show()
```

```python
sedan_gas_cars= df.loc[(df['fuel-type'] == 'gas') & (df['body-style'] == 'sedan')]
sedan_gas_counts = sedan_gas_cars['make'].value_counts()
sedan_gas_counts_df = sedan_gas_counts.to_frame().reset_index()
sedan_gas_counts_df= sedan_gas_counts_df.rename(columns={'index': 'Merk Mobil', 'make': 'Jumlah Mobil'})
sedan_gas_counts_df
```

```python
sedan_gas_counts.plot(kind='bar', color='blue')
plt.title("Jumlah Mobil dengan Bahan Bakar Gas dan Tipe Sedan")
plt.xlabel("Merk Mobil")
plt.ylabel("Jumlah Mobil")
plt.xticks(rotation=30)
plt.show()
```

```python
# Kelompokkan berdasarkan 'make' dan hitung rata-rata 'horsepower' untuk setiap merek
top_5_horsepower = df.groupby('make')['horsepower'].mean().nlargest(5)

# Konversi hasil ke dalam DataFrame
top_5_horsepower_df = top_5_horsepower.reset_index()

# Ubah nama kolom
top_5_horsepower_df = top_5_horsepower_df.rename(columns={'make': 'Merek Mobil'})

# Tampilkan DataFrame
top_5_horsepower_df
```

```python
ax = top_5_horsepower.plot(kind='bar', color ='pink')
plt.title ("5 Jenis Mobil Dengan Horsepower Terbesar")
plt.xlabel('Merk')
plt.ylabel('Horsepower')
plt.xticks(rotation=0)

for i, v in enumerate(top_5_horsepower):
    ax.text(i,v-3, str(v), ha='center', va='bottom')
```
